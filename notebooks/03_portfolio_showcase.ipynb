## User Workflow Demonstration

Let's walk through a typical user scenario that showcases the technical capabilities:

```python
# Import key modules to demonstrate the pipeline
import sys
sys.path.append('../src')

from utils.image_processor import extract_image_metadata, analyze_colors
from api_services.vision import BeeVisionAnalyzer
from api_services.weather import get_weather_open_meteo
import pandas as pd
import plotly.express as px
```

### Step 1: Automated Metadata Extraction

**Technical Challenge**: Reliably extract EXIF data from diverse camera formats while handling edge cases

**Implementation**: Multi-library approach with fallback mechanisms

```python
# Demo with sample image (replace with actual hive photo)
sample_image_path = "../data/uploads/sample_hive_photo.jpg"

# Extract comprehensive metadata
metadata = extract_image_metadata(sample_image_path)

print("📍 GPS Location:", metadata.get('gps_coordinates', 'Not available'))
print("📅 Date Taken:", metadata.get('date_taken', 'Not available'))
print("📱 Camera Model:", metadata.get('camera_model', 'Not available'))
print("⚙️ Technical Details:", f"{metadata.get('focal_length', 'N/A')} | f/{metadata.get('aperture', 'N/A')} | ISO {metadata.get('iso', 'N/A')}")
```

### Step 2: Computer Vision Analysis

**Technical Challenge**: Identify bee-specific features and hive health indicators from diverse photo conditions

**Implementation**: Google Cloud Vision API with custom confidence thresholds and bee-domain optimization

```python
# Initialize vision analyzer
vision_analyzer = BeeVisionAnalyzer()

# Analyze image for bee-related content
vision_results = vision_analyzer.analyze_image(sample_image_path)

print("🐝 Detected Objects:")
for obj in vision_results.get('objects', [])[:5]:  # Show top 5
    print(f"  - {obj['name']}: {obj['confidence']:.1%} confidence")

print("\n📊 Scene Analysis:")
for label in vision_results.get('labels', [])[:3]:  # Show top 3
    print(f"  - {label['description']}: {label['confidence']:.1%}")
```

### Step 3: Environmental Context Integration

**Technical Challenge**: Correlate photo timing/location with historical weather data

**Implementation**: Open-Meteo API integration with caching for performance

```python
# Get weather data for photo location and time
if metadata.get('gps_coordinates') and metadata.get('date_taken'):
    lat, lon = metadata['gps_coordinates']
    date_taken = metadata['date_taken']
    
    weather_data = get_weather_open_meteo(lat, lon, date_taken)
    
    print("🌤️ Environmental Conditions:")
    print(f"  Temperature: {weather_data.get('temperature', 'N/A')}°C")
    print(f"  Humidity: {weather_data.get('humidity', 'N/A')}%")
    print(f"  Wind Speed: {weather_data.get('wind_speed', 'N/A')} km/h")
    print(f"  Conditions: {weather_data.get('weather_description', 'N/A')}")
```

### Step 4: Color Analysis for Hive Health

**Technical Challenge**: Extract meaningful color patterns that correlate with hive health indicators

**Implementation**: ColorThief library with custom clustering for bee/honey-relevant color analysis

```python
# Analyze color palette
colors = analyze_colors(sample_image_path)

print("🎨 Dominant Color Palette:")
for i, color in enumerate(colors.get('palette', [])[:5]):
    hex_color = '#{:02x}{:02x}{:02x}'.format(*color)
    print(f"  Color {i+1}: {hex_color} - RGB{color}")

# Visualize color distribution
color_df = pd.DataFrame({
    'Color': [f'Color {i+1}' for i in range(len(colors.get('palette', [])))],
    'RGB': [str(color) for color in colors.get('palette', [])],
    'Hex': ['#{:02x}{:02x}{:02x}'.format(*color) for color in colors.get('palette', [])]
})

# Display color palette visualization here
print("\n📈 Color analysis complete - visualization would render in actual notebook")
```
